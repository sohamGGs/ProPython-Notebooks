```python
import threading
import time

def print_numbers():
    for i in range(5):
        time.sleep(1)
        print(f"Number: {i}")

def print_letters():
    for char in "ABCDE":
        time.sleep(1.5)
        print(f"Letter: {char}")

# Creating threads
t1 = threading.Thread(target=print_numbers)
t2 = threading.Thread(target=print_letters)

t1.start()
t2.start()

t1.join() # Wait for t1 to finish
t2.join() # Wait for t2 to finish
print("Done!")
```
