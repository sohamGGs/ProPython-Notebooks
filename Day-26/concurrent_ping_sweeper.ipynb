```python
import threading
import requests

def check_site(url):
    try:
        response = requests.get(url, timeout=5)
        print(f"[{url}] Status: {response.status_code}")
    except:
        print(f"[{url}] Status: DOWN")

sites = [
    "https://google.com", "https://github.com",
    "https://python.org", "https://stackoverflow.com"
]

threads = []
for site in sites:
    t = threading.Thread(target=check_site, args=(site,))
    threads.append(t)
    t.start()

for t in threads: t.join()
```
