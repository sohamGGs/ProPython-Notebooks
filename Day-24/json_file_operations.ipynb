```python
import json
import os

filename = "config.json"
user_settings = {"theme": "dark", "notifications": False, "font_size": 14}

# Writing to a file
with open(filename, "w") as f:
    json.dump(user_settings, f, indent=4)

# Reading from a file
if os.path.exists(filename):
    with open(filename, "r") as f:
        loaded_settings = json.load(f)
        print(f"Theme loaded: {loaded_settings.get('theme')}")
```
