```python
import json
from datetime import datetime

class CustomEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, datetime):
            return obj.isoformat()
        return super().default(obj)

event = {"title": "Python Exam", "date": datetime.now()}
# This would fail without the 'cls' argument
encoded_event = json.dumps(event, cls=CustomEncoder)
print(f"Encoded with Date: {encoded_event}")
```
