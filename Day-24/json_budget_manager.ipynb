```python
import json
import os

class BudgetManager:
    def __init__(self, filename="budget.json"):
        self.filename = filename
        self.data = self._load_data()

    def _load_data(self):
        if os.path.exists(self.filename):
            with open(self.filename, "r") as f:
                return json.load(f)
        return {"total_budget": 0, "expenses": []}

    def add_expense(self, item, amount):
        self.data["expenses"].append({"item": item, "amount": amount})
        self._save_data()
        print(f"Logged {item}: ₹{amount}")

    def _save_data(self):
        with open(self.filename, "w") as f:
            json.dump(self.data, f, indent=4)

if __name__ == "__main__":
    manager = BudgetManager()
    manager.add_expense("Laptop Stand", 1200)
    print(f"Total entries: {len(manager.data['expenses'])}")
```
