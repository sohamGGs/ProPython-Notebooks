```python
import json

data = {
    "name": "Soham Patil",
    "college": "GHRCEM",
    "skills": ["Python", "DSA", "Web Dev"],
    "is_student": True
}

# Serialize (Convert Python dict to JSON string)
json_string = json.dumps(data, indent=4)
print(f"JSON String:\n{json_string}")

# Deserialize (Convert JSON string back to Python dict)
new_data = json.loads(json_string)
print(f"Name from Dict: {new_data['name']}")
```
