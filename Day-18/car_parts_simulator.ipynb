```python
class Component:
    def __init__(self, name: str, health: int):
        self.name = name
        self.health = health

class Vehicle:
    def __init__(self, name: str):
        self.name = name
        self.parts = []

    def add_part(self, part: Component):
        self.parts.append(part)

    def check_health(self):
        print(f"--- {self.name} Status ---")
        for p in self.parts:
            status = "OK" if p.health > 50 else "SERVICE NEEDED"
            print(f"{p.name}: {status}")

if __name__ == "__main__":
    car = Vehicle("Mahindra XUV")
    car.add_part(Component("Tires", 80))
    car.add_part(Component("Battery", 30))
    car.check_health()
```
