```python
class Writer:
    def write(self):
        print("Writing content...")

class Developer:
    def __init__(self, writer_role: Writer):
        self.writer = writer_role

    def create_docs(self):
        # Delegating the writing task to the Writer object
        self.writer.write()

if __name__ == "__main__":
    dev = Developer(Writer())
    dev.create_docs()
```
