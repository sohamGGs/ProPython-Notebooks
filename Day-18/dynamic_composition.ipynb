```python
class FlyingBehavior:
    def move(self): print("I am flying high!")

class WalkingBehavior:
    def move(self): print("I am walking on the ground.")

class Robot:
    def __init__(self, behavior):
        self.behavior = behavior

    def execute_move(self):
        self.behavior.move()

if __name__ == "__main__":
    # We can swap behaviors at runtime!
    r = Robot(WalkingBehavior())
    r.execute_move()
    r.behavior = FlyingBehavior()
    r.execute_move()
```
