```python
class Engine:
    def start(self):
        return "Engine is roaring!"

class Car:
    def __init__(self, model: str):
        self.model = model
        # Composition: Car 'has an' Engine
        self.engine = Engine()

    def start_car(self):
        return f"{self.model}: {self.engine.start()}"

if __name__ == "__main__":
    my_car = Car("Tesla")
    print(my_car.start_car())
```
