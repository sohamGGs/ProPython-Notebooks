```python
# --- STEP 1: Creating a 'virtual' package structure ---
with open("math_utils.py", "w") as f:
    f.write('''
def add(a, b): return a + b
def subtract(a, b): return a - b
def multiply(a, b): return a * b
''')

with open("advanced_utils.py", "w") as f:
    f.write('''
def power(a, b): return a ** b
''')

# --- STEP 2: Using the modules ---
import math_utils
from advanced_utils import power

def run_calculator():
    print("--- Multi-file Calculator Library ---")
    x, y = 10, 5
    print(f"{x} + {y} = {math_utils.add(x, y)}")
    print(f"{x} to the power of {y} = {power(x, y)}")

if __name__ == "__main__":
    run_calculator()
```
