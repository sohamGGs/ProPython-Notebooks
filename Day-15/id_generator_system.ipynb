```python
class User:
    _id_counter = 1000  # Private class attribute

    def __init__(self, username: str):
        self.username = username
        self.user_id = User._generate_id()

    @classmethod
    def _generate_id(cls) -> int:
        cls._id_counter += 1
        return cls._id_counter

if __name__ == "__main__":
    u1 = User("Soham_Patil")
    u2 = User("Vedant_Khaire")
    print(f"User: {u1.username}, ID: {u1.user_id}")
    print(f"User: {u2.username}, ID: {u2.user_id}")
```
