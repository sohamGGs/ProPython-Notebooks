```python
class Robot:
    population = 0  # Class attribute

    def __init__(self, name: str):
        self.name = name
        Robot.update_population()

    @classmethod
    def update_population(cls):
        cls.population += 1

    @staticmethod
    def calculate_battery(voltage: float, current: float) -> float:
        return voltage * current  # Doesn't need 'self' or 'cls'

if __name__ == "__main__":
    r1 = Robot("Alpha")
    r2 = Robot("Beta")
    print(f"Total Robots: {Robot.population}")
    print(f"Power Draw: {Robot.calculate_battery(12, 1.5)}W")
```
