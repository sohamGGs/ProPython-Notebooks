```python
class A:
    def greet(self): print("Hello from A")

class B(A):
    def greet(self): print("Hello from B")

class C(A):
    def greet(self): print("Hello from C")

class D(B, C):
    pass

if __name__ == "__main__":
    obj = D()
    obj.greet() # Which one will it pick?
    print(f"Order: {D.mro()}") # MRO follows the C3 Linearization algorithm
```
