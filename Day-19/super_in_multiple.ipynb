```python
class Base:
    def __init__(self):
        print("Base Init")

class Left(Base):
    def __init__(self):
        super().__init__()
        print("Left Init")

class Right(Base):
    def __init__(self):
        super().__init__()
        print("Right Init")

class Combined(Left, Right):
    def __init__(self):
        super().__init__() # This triggers the entire MRO chain!
        print("Combined Init")

if __name__ == "__main__":
    c = Combined()
```
