```python
class Warrior:
    def attack(self): return "Sword Slash!"

class Mage:
    def cast_spell(self): return "Fireball!"

class Paladin(Warrior, Mage):
    def special_move(self):
        return f"Combo: {self.attack()} and {self.cast_spell()}"

if __name__ == "__main__":
    player = Paladin()
    print(f"Character Status: {player.special_move()}")
```
