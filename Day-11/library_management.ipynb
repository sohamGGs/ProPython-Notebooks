```python
from typing import List

class Book:
    def __init__(self, title: str, author: str):
        self.title = title
        self.author = author
        self.is_available = True

class Library:
    def __init__(self):
        self.books: List[Book] = []

    def add_book(self, title: str, author: str):
        new_book = Book(title, author)
        self.books.append(new_book)
        return f"Added: {title} by {author}"

    def show_available(self):
        available = [b.title for b in self.books if b.is_available]
        return f"Available Books: {available}"

if __name__ == "__main__":
    my_lib = Library()
    print(my_lib.add_book("Python Crash Course", "Eric Matthes"))
    print(my_lib.add_book("Clean Code", "Robert C. Martin"))
    print(my_lib.show_available())
```
