```python
from typing import List

class Student:
    # Class Attribute (Shared by all students)
    college = "GHRCEM, Pune"

    def __init__(self, name: str, branch: str):
        # Instance Attributes (Unique to each object)
        self.name = name
        self.branch = branch

    def display_info(self) -> None:
        print(f"Name: {self.name} | Branch: {self.branch} | College: {self.college}")

# Creating objects
student1 = Student("Soham", "CSE")
student2 = Student("Vedant", "Data Science")

student1.display_info()
student2.display_info()
```
