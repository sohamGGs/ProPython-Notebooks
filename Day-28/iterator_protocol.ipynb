```python
class MyCounter:
    def __init__(self, low: int, high: int):
        self.current = low
        self.high = high

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.high:
            raise StopIteration
        else:
            self.current += 1
            return self.current - 1

if __name__ == "__main__":
    counter = MyCounter(1, 5)
    for num in counter:
        print(f"Count: {num}")
```
