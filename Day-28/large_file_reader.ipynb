```python
import os

def read_large_file(file_path):
    """Generator to read a file line by line."""
    with open(file_path, "r") as file:
        for line in file:
            yield line.strip()

if __name__ == "__main__":
    # Create a dummy large-ish file
    with open("big_data.txt", "w") as f:
        for i in range(1000):
            f.write(f"Line number {i} with some data\n")

    # Use the generator
    data_gen = read_large_file("big_data.txt")

    print("First 3 lines processed via generator:")
    print(next(data_gen))
    print(next(data_gen))
    print(next(data_gen))

    # Cleanup
    os.remove("big_data.txt")
```
