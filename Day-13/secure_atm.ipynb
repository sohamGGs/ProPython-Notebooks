```python
class ATM:
    def __init__(self, pin: str, balance: float):
        self.__pin = pin
        self.__balance = balance

    def withdraw(self, entered_pin: str, amount: float):
        if entered_pin != self.__pin:
            return "Error: Invalid PIN!"
        if amount > self.__balance:
            return "Error: Insufficient funds!"

        self.__balance -= amount
        return f"Withdrawal successful! Remaining: ₹{self.__balance}"

if __name__ == "__main__":
    my_atm = ATM("1234", 5000.0)
    print(my_atm.withdraw("1111", 500))  # Wrong PIN
    print(my_atm.withdraw("1234", 1000)) # Correct PIN
```
