```python
from multiprocessing import Process, Queue

def producer(q):
    q.put("Data from Producer")

def consumer(q):
    data = q.get()
    print(f"Consumer received: {data}")

if __name__ == "__main__":
    q = Queue()
    p1 = Process(target=producer, args=(q,))
    p2 = Process(target=consumer, args=(q,))
    p1.start()
    p2.start()
    p1.join()
    p2.join()
```
