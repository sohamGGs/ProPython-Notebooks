```python
from datetime import datetime, timedelta

class StreakTracker:
    def __init__(self, habit_name: str):
        self.habit = habit_name
        self.history = []

    def log_completion(self, date_str: str):
        date_obj = datetime.strptime(date_str, "%Y-%m-%d").date()
        self.history.append(date_obj)
        print(f"Logged {self.habit} for {date_obj}")

    def check_streak(self):
        if not self.history: return 0
        sorted_dates = sorted(self.history, reverse=True)
        streak = 1
        for i in range(len(sorted_dates)-1):
            if sorted_dates[i] - sorted_dates[i+1] == timedelta(days=1):
                streak += 1
            else:
                break
        return streak

if __name__ == "__main__":
    tracker = StreakTracker("Coding")
    tracker.log_completion("2026-03-07")
    tracker.log_completion("2026-03-06")
    print(f"Current Streak: {tracker.check_streak()} days!")
```
