```python
from datetime import datetime
import zoneinfo # Available in modern Python (Colab 3.9+)

try:
    utc_time = datetime.now(zoneinfo.ZoneInfo("UTC"))
    ist_time = datetime.now(zoneinfo.ZoneInfo("Asia/Kolkata"))
    print(f"UTC: {utc_time}")
    print(f"IST (Pune): {ist_time}")
except:
    print("Zoneinfo not available, using simple offset.")
```
