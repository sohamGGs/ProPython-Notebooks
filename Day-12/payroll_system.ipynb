```python
class Worker:
    def __init__(self, name: str):
        self.name = name

    def calculate_pay(self) -> float:
        return 0.0

class FullTimeWorker(Worker):
    def calculate_pay(self) -> float:
        return 50000.0  # Fixed monthly salary

class PartTimeWorker(Worker):
    def __init__(self, name: str, hours: int, rate: float):
        super().__init__(name)
        self.hours = hours
        self.rate = rate

    def calculate_pay(self) -> float:
        return self.hours * self.rate

if __name__ == "__main__":
    workers = [FullTimeWorker("Soham"), PartTimeWorker("Ved", 20, 500)]
    for w in workers:
        print(f"{w.name} Pay: ₹{w.calculate_pay()}")
```
