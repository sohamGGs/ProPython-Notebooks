```python
class Account:
    def __init__(self, owner: str, balance: float):
        self.owner = owner
        self._balance = balance
        self._logs = []

    @property
    def balance(self):
        return f"Account Balance: ₹{self._balance}"

    @balance.setter
    def balance(self, amount: float):
        change = amount - self._balance
        self._logs.append(f"Changed by: {change}")
        self._balance = amount

if __name__ == "__main__":
    acc = Account("Soham Patil", 1000)
    acc.balance = 1500  # Uses setter
    acc.balance = 1200  # Uses setter
    print(acc.balance)
    print(f"History: {acc._logs}")
```
