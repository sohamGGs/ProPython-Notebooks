```python
class Temperature:
    def __init__(self, celsius: float):
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value: float):
        """The Setter: Validates data before saving."""
        if value < -273.15:
            raise ValueError("Temperature below absolute zero is impossible!")
        self._celsius = value

if __name__ == "__main__":
    t = Temperature(25)
    t.celsius = 30  # Works fine
    print(f"Current Temp: {t.celsius}°C")
```
