```python
class Rectangle:
    def __init__(self, width: float, height: float):
        self.width = width
        self.height = height

    @property
    def area(self) -> float:
        """Computed property: No need to store 'area' separately."""
        return self.width * self.height

if __name__ == "__main__":
    r = Rectangle(10, 5)
    print(f"Area: {r.area}")
    r.width = 20
    print(f"New Area: {r.area}") # Automatically updates!
```
