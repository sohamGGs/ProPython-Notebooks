```python
class Employee:
    def __init__(self, name: str, salary: float):
        self.name = name
        self._salary = salary  # Protected attribute

    @property
    def salary(self) -> str:
        """The Getter: Formats the salary for display."""
        return f"₹{self._salary:,.2f}"

if __name__ == "__main__":
    emp = Employee("Soham", 75000)
    print(f"Name: {emp.name} | Salary: {emp.salary}")
```
