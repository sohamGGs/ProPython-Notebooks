```python
class Bird:
    def fly(self):
        print("Most birds can fly.")

class Sparrow(Bird):
    def fly(self):
        print("Sparrows fly high.")

class Ostrich(Bird):
    def fly(self):
        print("Ostriches cannot fly, they run!")

def make_it_fly(bird_obj: Bird):
    bird_obj.fly()

if __name__ == "__main__":
    make_it_fly(Sparrow())
    make_it_fly(Ostrich())
```
