```python
from abc import ABC, abstractmethod

class Payment(ABC):
    @abstractmethod
    def process(self, amount: float):
        pass

class UPIPayment(Payment):
    def process(self, amount: float):
        print(f"Processing UPI payment of ₹{amount} via PhonePe/GPay.")

class CardPayment(Payment):
    def process(self, amount: float):
        print(f"Processing Card payment of ₹{amount} (Visa/Mastercard).")

if __name__ == "__main__":
    payments = [UPIPayment(), CardPayment()]
    for p in payments:
        p.process(499.0)
```
