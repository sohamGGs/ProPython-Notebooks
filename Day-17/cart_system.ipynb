```python
class Item:
    def __init__(self, name: str, price: float):
        self.name, self.price = name, price

class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, item: Item):
        self.items.append(item)

    def __add__(self, other_item: Item):
        """Shortcut to add items: cart + item"""
        self.add_item(other_item)
        return self

    def __gt__(self, other):
        """Compare two carts: cart1 > cart2"""
        return sum(i.price for i in self.items) > sum(i.price for i in other.items)

if __name__ == "__main__":
    cart1 = ShoppingCart()
    cart1 = cart1 + Item("Laptop", 60000) # Using __add__

    cart2 = ShoppingCart()
    cart2 = cart2 + Item("Phone", 20000)

    print(f"Is Cart 1 more expensive? {cart1 > cart2}")
```
