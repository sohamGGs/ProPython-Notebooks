```python
class Car:
    def __init__(self, model: str, year: int):
        self.model = model
        self.year = year

    def __str__(self):
        """For end users: Friendly string."""
        return f"{self.model} ({self.year})"

    def __repr__(self):
        """For developers: Unambiguous representation."""
        return f"Car(model='{self.model}', year={self.year})"

if __name__ == "__main__":
    my_car = Car("Tesla Model 3", 2024)
    print(str(my_car))
    print(repr(my_car))
```
