```python
class Vector:
    def __init__(self, x: int, y: int):
        self.x, self.y = x, y

    def __add__(self, other):
        """Allows using the '+' operator."""
        return Vector(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

if __name__ == "__main__":
    v1 = Vector(2, 4)
    v2 = Vector(1, 3)
    print(v1 + v2) # Calls __add__
```
